```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

print("TensorFlow sürümü:", tf.__version__)
```

```python
from tensorflow.keras.datasets import mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()

print("Eğitim görüntüleri:", x_train.shape)
print("Test görüntüleri:", x_test.shape)
print("Eğitim etiketleri:", y_train.shape)
print("Test etiketleri:", y_test.shape)
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Minimum:", x_train.min())
print("Maximum:", x_train.max())
```

```python
plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(f"Rakam: {y_train[i]}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

print("x_train:", x_train.shape)
print("x_test:", x_test.shape)
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, Dropout

model = Sequential([
    Input(shape=(28, 28, 1)),

    Conv2D(32, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation="relu"),
    Dropout(0.5),

    Dense(10, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
import tensorflow as tf

with tf.GradientTape() as tape:
    predictions = model(x_train[:32], training=True)
    loss = tf.keras.losses.sparse_categorical_crossentropy(
        y_train[:32],
        predictions
    )
    loss = tf.reduce_mean(loss)

gradients = tape.gradient(loss, model.trainable_weights)

for weight, gradient in zip(model.trainable_weights, gradients):
    if gradient is None:
        print(weight.name, "→ GRADIENT YOK")
    else:
        print(
            weight.name,
            "→ gradient ortalaması:",
            tf.reduce_mean(tf.abs(gradient)).numpy()
        )
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, LeakyReLU
from tensorflow.keras.optimizers import Adam

model = Sequential([
    Input(shape=(28, 28, 1)),

    Conv2D(32, (3, 3), padding="same", kernel_initializer="he_normal"),
    LeakyReLU(negative_slope=0.01),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), padding="same", kernel_initializer="he_normal"),
    LeakyReLU(negative_slope=0.01),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, kernel_initializer="he_normal"),
    LeakyReLU(negative_slope=0.01),

    Dense(10, activation="softmax")
])

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# PIL görüntüsünü numpy dizisine çevir
crop_array = np.array(crop)

# Hafif bulanıklaştır (ışık/gölge etkisini azaltır)
blur = cv2.GaussianBlur(crop_array, (5, 5), 0)

# Adaptif eşikleme
binary = cv2.adaptiveThreshold(
    blur,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY_INV,
    31,
    10
)

# Sonucu göster
plt.figure(figsize=(12, 4))
plt.imshow(binary, cmap="gray")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Konturları bul
contours, _ = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

# Soldan sağa sıralamak için kutuları oluştur
boxes = []

for cnt in contours:
    x, y, w, h = cv2.boundingRect(cnt)

    # Çok küçük lekeleri alma
    if w > 5 and h > 15:
        boxes.append((x, y, w, h))

# Soldan sağa sırala
boxes = sorted(boxes, key=lambda b: b[0])

# Bulunan parçaları göster
plt.figure(figsize=(15, 3))

for i, (x, y, w, h) in enumerate(boxes):
    digit = binary[y:y+h, x:x+w]

    plt.subplot(1, len(boxes), i+1)
    plt.imshow(digit, cmap="gray")
    plt.axis("off")

plt.show()

print("Bulunan parça sayısı:", len(boxes))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Her sütunda kaç beyaz piksel var?
vertical_sum = np.sum(binary > 0, axis=0)

# Yazı bulunan sütunları bul
active = vertical_sum > 3

# Başlangıç-bitiş aralıklarını çıkar
segments = []
start = None

for i, val in enumerate(active):
    if val and start is None:
        start = i
    elif not val and start is not None:
        end = i
        if end - start > 5:
            segments.append((start, end))
        start = None

if start is not None:
    segments.append((start, len(active)))

print("Bulunan segment sayısı:", len(segments))

# Segmentleri göster
plt.figure(figsize=(15, 3))

for i, (x1, x2) in enumerate(segments):
    part = binary[:, x1:x2]

    plt.subplot(1, len(segments), i + 1)
    plt.imshow(part, cmap="gray")
    plt.axis("off")

plt.show()
```

```python
# İlk iki rakam: 0, 6
# 3. parça "/" olduğu için almıyoruz
# Sonra: 0, 9
# 6. parça "/" olduğu için almıyoruz

digit_parts = []

# 0 ve 6
for idx in [0, 1]:
    x1, x2 = segments[idx]
    digit_parts.append(binary[:, x1:x2])

# 0 ve 9
for idx in [3, 4]:
    x1, x2 = segments[idx]
    digit_parts.append(binary[:, x1:x2])

# "20" parçasını ikiye böl
x1, x2 = segments[6]
part20 = binary[:, x1:x2]

middle = part20.shape[1] // 2
digit_parts.append(part20[:, :middle])
digit_parts.append(part20[:, middle:])

# "26" parçasını ikiye böl
x1, x2 = segments[7]
part26 = binary[:, x1:x2]

middle = part26.shape[1] // 2
digit_parts.append(part26[:, :middle])
digit_parts.append(part26[:, middle:])

# Sonucu göster
plt.figure(figsize=(12, 3))

for i, digit in enumerate(digit_parts):
    plt.subplot(1, 8, i + 1)
    plt.imshow(digit, cmap="gray")
    plt.axis("off")
    plt.title(str(i + 1))

plt.show()

print("Rakam sayısı:", len(digit_parts))
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

processed_digits = []

for digit in digit_parts:
    # Sadece rakamın olduğu satırları bul
    rows = np.where(np.sum(digit > 0, axis=1) > 0)[0]
    cols = np.where(np.sum(digit > 0, axis=0) > 0)[0]

    if len(rows) == 0 or len(cols) == 0:
        continue

    # Rakamı sıkı şekilde kırp
    d = digit[rows[0]:rows[-1]+1, cols[0]:cols[-1]+1]

    # 20x20 içine oranı bozmadan küçült
    h, w = d.shape

    if h > w:
        new_h = 20
        new_w = max(1, int(w * (20 / h)))
    else:
        new_w = 20
        new_h = max(1, int(h * (20 / w)))

    resized = cv2.resize(
        d.astype("uint8"),
        (new_w, new_h),
        interpolation=cv2.INTER_AREA
    )

    # 28x28 siyah zemin oluştur
    canvas = np.zeros((28, 28), dtype=np.uint8)

    # Ortala
    y_offset = (28 - new_h) // 2
    x_offset = (28 - new_w) // 2

    canvas[
        y_offset:y_offset+new_h,
        x_offset:x_offset+new_w
    ] = resized

    processed_digits.append(canvas)

# Hepsini göster
plt.figure(figsize=(12, 3))

for i, d in enumerate(processed_digits):
    plt.subplot(1, 8, i + 1)
    plt.imshow(d, cmap="gray")
    plt.axis("off")

plt.show()

print("Hazırlanan rakam sayısı:", len(processed_digits))
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist

# MNIST verisini yükle
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize et
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Kanal boyutu ekle
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# CNN modeli
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation="relu", input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def split_two_digits(part):
    # Her sütundaki beyaz piksel sayısı
    col_sum = np.sum(part > 0, axis=0)

    width = part.shape[1]

    # Kenarları dikkate alma, orta bölgeyi ara
    start = int(width * 0.30)
    end = int(width * 0.70)

    # İki rakam arasındaki en boş sütunu bul
    split_point = start + np.argmin(col_sum[start:end])

    left = part[:, :split_point]
    right = part[:, split_point:]

    return left, right


# Yeni rakam listesi
digit_parts_fixed = []

# 0, 6
for idx in [0, 1]:
    x1, x2 = segments[idx]
    digit_parts_fixed.append(binary[:, x1:x2])

# 0, 9
for idx in [3, 4]:
    x1, x2 = segments[idx]
    digit_parts_fixed.append(binary[:, x1:x2])

# 20'yi düzgün ayır
x1, x2 = segments[6]
part20 = binary[:, x1:x2]

digit2, digit0 = split_two_digits(part20)

digit_parts_fixed.append(digit2)
digit_parts_fixed.append(digit0)

# 26'yı düzgün ayır
x1, x2 = segments[7]
part26 = binary[:, x1:x2]

digit2, digit6 = split_two_digits(part26)

digit_parts_fixed.append(digit2)
digit_parts_fixed.append(digit6)

# Sonucu göster
plt.figure(figsize=(12, 3))

for i, d in enumerate(digit_parts_fixed):
    plt.subplot(1, 8, i + 1)
    plt.imshow(d, cmap="gray")
    plt.axis("off")
    plt.title(str(i + 1))

plt.tight_layout()
plt.show()

print("Rakam sayısı:", len(digit_parts_fixed))
```

```python
from google.colab import files
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import cv2

uploaded = files.upload()

filename = list(uploaded.keys())[0]
img_student = Image.open(filename).convert("L")

plt.figure(figsize=(12,4))
plt.imshow(img_student, cmap="gray")
plt.axis("off")
plt.show()
```

```python
w, h = img_student.size

student_crop = img_student.crop((
    int(w * 0.42),
    int(h * 0.25),
    int(w * 0.95),
    int(h * 0.75)
))

plt.figure(figsize=(12,4))
plt.imshow(student_crop, cmap="gray")
plt.axis("off")
plt.show()
```

```python
# Öğrenci numarası görüntüsünü siyah-beyaz yap
student_array = np.array(student_crop)

blur = cv2.GaussianBlur(student_array, (5, 5), 0)

student_binary = cv2.adaptiveThreshold(
    blur,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY_INV,
    31,
    10
)

plt.figure(figsize=(12, 4))
plt.imshow(student_binary, cmap="gray")
plt.axis("off")
plt.show()
```

```python
# Her sütundaki beyaz piksel miktarını hesapla
vertical_sum = np.sum(student_binary > 0, axis=0)

# Yazı bulunan sütunları belirle
active = vertical_sum > 3

student_segments = []
start = None

for i, val in enumerate(active):

    if val and start is None:
        start = i

    elif not val and start is not None:
        end = i

        if end - start > 5:
            student_segments.append((start, end))

        start = None

if start is not None:
    student_segments.append((start, len(active)))

print("Bulunan rakam/parça sayısı:", len(student_segments))

# Bulunan parçaları göster
plt.figure(figsize=(15, 3))

for i, (x1, x2) in enumerate(student_segments):

    digit = student_binary[:, x1:x2]

    plt.subplot(1, len(student_segments), i + 1)
    plt.imshow(digit, cmap="gray")
    plt.title(str(i + 1))
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
processed_student_digits = []

for x1, x2 in student_segments:

    digit = student_binary[:, x1:x2]

    rows = np.where(np.sum(digit > 0, axis=1) > 0)[0]
    cols = np.where(np.sum(digit > 0, axis=0) > 0)[0]

    if len(rows) == 0 or len(cols) == 0:
        continue

    d = digit[
        rows[0]:rows[-1] + 1,
        cols[0]:cols[-1] + 1
    ]

    # Hafif kalınlaştır
    kernel = np.ones((2, 2), np.uint8)
    d = cv2.dilate(d.astype("uint8"), kernel, iterations=1)

    h, w = d.shape

    if h > w:
        new_h = 18
        new_w = max(1, int(w * 18 / h))
    else:
        new_w = 18
        new_h = max(1, int(h * 18 / w))

    resized = cv2.resize(
        d,
        (new_w, new_h),
        interpolation=cv2.INTER_AREA
    )

    canvas = np.zeros((28, 28), dtype=np.uint8)

    y = (28 - new_h) // 2
    x = (28 - new_w) // 2

    canvas[y:y+new_h, x:x+new_w] = resized

    processed_student_digits.append(canvas)

# Göster
plt.figure(figsize=(15, 3))

for i, d in enumerate(processed_student_digits):
    plt.subplot(1, len(processed_student_digits), i + 1)
    plt.imshow(d, cmap="gray")
    plt.axis("off")

plt.tight_layout()
plt.show()

print("Hazırlanan rakam sayısı:", len(processed_student_digits))
```
